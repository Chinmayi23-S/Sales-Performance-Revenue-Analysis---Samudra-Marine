```python
import os

# Folder containing CSVs
csv_folder = "C:/temp"

# Map CSV filenames to table names (excluding orders since it's already loaded)
csv_map = {
    "Country_Code.csv": "country_code",
    "Customers.csv": "customers",
    "Customers_Segment.csv": "customers_segment",
    "Ports.csv": "ports",
    "Product Category.csv": "product_category",
    "Products.csv": "products",
    "Sales Region.csv": "sales_region",
    "Sales Representative.csv": "sales_representative",
    "vessels.csv": "vessels"
}

# Loop through each CSV and push to PostgreSQL
for file, table in csv_map.items():
    path = os.path.join(csv_folder, file)
    try:
        df = pd.read_csv(path)
        df.to_sql(table, engine, if_exists="replace", index=False)
        print(f"✅ {table} table created and data loaded!")
    except Exception as e:
        print(f"❌ Failed to load {table}: {e}")
```
